```python
import os
import matplotlib.pyplot as plt
import pandas as pd


plt.style.use('seaborn-v0_8-paper')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['axes.linewidth'] = 1.0


COLOR_SF = '#1E3A8A'  # Deep Space Blue (Spaceflight)
COLOR_GC = '#0D9488'  # Slate Teal (Ground Control)


def generate_visuals():
    df['Total Mice (n)'] = df['Spaceflight (SF)'] + df['Ground Control (GC)']
    df['Label'] = df['Dataset ID'] + '\n(' + df['Mission'] + ')'

    os.makedirs('notebooks', exist_ok=True)

    
    # FIGURE 1: Sample Distribution by Dataset
    
    fig1, ax1 = plt.subplots(figsize=(6.5, 4.8), dpi=300)

    bars_sf = ax1.bar(
        df['Label'],
        df['Spaceflight (SF)'],
        label='Spaceflight (SF)',
        color=COLOR_SF,
        edgecolor='black',
        lw=0.7,
        width=0.55,
    )
    bars_gc = ax1.bar(
        df['Label'],
        df['Ground Control (GC)'],
        bottom=df['Spaceflight (SF)'],
        label='Ground Control (GC)',
        color=COLOR_GC,
        edgecolor='black',
        lw=0.7,
        width=0.55,
    )

    for i in range(len(df)):
        sf = df['Spaceflight (SF)'][i]
        gc = df['Ground Control (GC)'][i]
        total = df['Total Mice (n)'][i]

        ax1.text(
            i,
            sf / 2,
            str(sf),
            ha='center',
            va='center',
            color='white',
            weight='bold',
            fontsize=8.5,
        )
        ax1.text(
            i,
            sf + gc / 2,
            str(gc),
            ha='center',
            va='center',
            color='white',
            weight='bold',
            fontsize=8.5,
        )

        ax1.text(
            i,
            total + 0.8,
            f'n={total}',
            ha='center',
            va='bottom',
            color='#1F2937',
            weight='bold',
            fontsize=8.5,
        )

    ax1.set_ylabel(
        'Biological Replicates Count (n)', fontsize=10, weight='bold'
    )
    ax1.set_title(
        'Multi-Omics Sample Distribution Matrix (N=79)',
        fontsize=11,
        weight='bold',
        loc='left',
        pad=12,
    )

    ax1.set_ylim(0, max(df['Total Mice (n)']) + 5)

    ax1.spines['top'].set_visible(False)
    ax1.spines['right'].set_visible(False)
    ax1.yaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax1.set_axisbelow(True)

    ax1.legend(
        frameon=True,
        facecolor='white',
        framealpha=0.9,
        edgecolor='#E5E7EB',
        loc='upper right',
        fontsize=9,
    )

    plt.tight_layout()
    plt.savefig(
        'notebooks/figure1_sample_distribution.png', bbox_inches='tight'
    )
    plt.close()

    
    # FIGURE 2: Biological Sample Volume Grouped by Omics Layer
    
    fig2, ax2 = plt.subplots(figsize=(6.5, 4.8), dpi=300)
    omics_summary = df.groupby('Omics Layer')[
        ['Spaceflight (SF)', 'Ground Control (GC)']
    ].sum()

    x = range(len(omics_summary))
    width = 0.32

    b1 = ax2.bar(
        [p - width / 2 for p in x],
        omics_summary['Spaceflight (SF)'],
        width,
        label='Spaceflight (SF)',
        color=COLOR_SF,
        edgecolor='black',
        lw=0.7,
    )
    b2 = ax2.bar(
        [p + width / 2 for p in x],
        omics_summary['Ground Control (GC)'],
        width,
        label='Ground Control (GC)',
        color=COLOR_GC,
        edgecolor='black',
        lw=0.7,
    )

    
    ax2.bar_label(
        b1, fmt='%d', padding=3, fontsize=9, weight='bold', color='#1F2937'
    )
    ax2.bar_label(
        b2, fmt='%d', padding=3, fontsize=9, weight='bold', color='#1F2937'
    )

    ax2.set_xticks(x)
    ax2.set_xticklabels(omics_summary.index, fontsize=10, weight='bold')
    ax2.set_ylabel('Total Pooled Replicates (n)', fontsize=10, weight='bold')
    ax2.set_title(
        'Biological Sample Volume Grouped by Omics Layer',
        fontsize=11,
        weight='bold',
        loc='left',
        pad=12,
    )

    
    max_val = omics_summary.values.max()
    ax2.set_ylim(0, max_val + 6)

    ax2.spines['top'].set_visible(False)
    ax2.spines['right'].set_visible(False)
    ax2.yaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax2.set_axisbelow(True)


    ax2.legend(
        frameon=True,
        facecolor='white',
        framealpha=0.9,
        edgecolor='#E5E7EB',
        loc='lower right',
        fontsize=9,
    )

    plt.tight_layout()
    plt.savefig('notebooks/figure2_omics_volume.png', bbox_inches='tight')
    plt.close()


generate_visuals()
```

```python
import os
import matplotlib.pyplot as plt
import pandas as pd

plt.style.use('seaborn-v0_8-paper')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['axes.linewidth'] = 1.0

COLOR_SF = '#1E3A8A'  # Deep Space Blue
COLOR_GC = '#0D9488'  # Slate Teal
COLOR_LINE = '#94A3B8'  # Slate Gray for connectors


def generate_age_visuals():
    df = df_age.copy()
    df['Label'] = df['Dataset ID'] + '\n(' + df['Mission'] + ')'
    os.makedirs('notebooks', exist_ok=True)

    
    # FIGURE 3: Rodent Lifespan Development Timeline Across Payloads
    
    fig3, ax3 = plt.subplots(figsize=(6.8, 4.8), dpi=300)

    for i in range(len(df)):
        launch_age = df['Age at Launch (wks)'][i]
        dissect_age = df['Age at Dissection (wks)'][i]

    
        ax3.plot(
            [launch_age, dissect_age],
            [i, i],
            color=COLOR_LINE,
            lw=2.2,
            zorder=1,
        )

  
        ax3.scatter(
            launch_age,
            i,
            color=COLOR_GC,
            s=50,
            label='Launch Baseline' if i == 0 else '',
            zorder=3,
        )
        ax3.scatter(
            dissect_age,
            i,
            color=COLOR_SF,
            s=60,
            marker='X',
            label='Dissection Terminal' if i == 0 else '',
            zorder=3,
        )

    
        ax3.text(
            launch_age - 0.6,
            i,
            f'{launch_age:.1f}',
            ha='right',
            va='center',
            fontsize=7.5,
            weight='bold',
            color='#374151',
        )

       
        ax3.text(
            dissect_age + 0.6,
            i,
            f'{dissect_age:.1f}',
            ha='left',
            va='center',
            fontsize=7.5,
            weight='bold',
            color=COLOR_SF,
        )

    ax3.set_yticks(range(len(df)))
    ax3.set_yticklabels(df['Label'], fontsize=8.5)
    ax3.set_xlabel(
        'Chronological Animal Age (Weeks)', fontsize=10, weight='bold'
    )
    ax3.set_title(
        'Rodent Lifespan Development Timeline Across Payloads',
        fontsize=11,
        weight='bold',
        loc='left',
        pad=12,
    )

  
    x_min = df['Age at Launch (wks)'].min() - 3.0
    x_max = df['Age at Dissection (wks)'].max() + 3.0
    ax3.set_xlim(x_min, x_max)

    ax3.spines['top'].set_visible(False)
    ax3.spines['right'].set_visible(False)
    ax3.xaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax3.set_axisbelow(True)

    ax3.legend(
        frameon=True,
        facecolor='white',
        framealpha=0.9,
        edgecolor='#E5E7EB',
        loc='lower right',
        fontsize=8.5,
    )

    plt.tight_layout()
    plt.savefig('notebooks/figure3_age_timeline.png', bbox_inches='tight')
    plt.close()

    
    # FIGURE 4: Spaceflight Sample Allocation by Biological Age Cohort
    
    fig4, ax4 = plt.subplots(figsize=(6.0, 4.8), dpi=300)

    def bin_age(age):
        if age < 16:
            return 'Young Adult\n(<16 wks)'
        elif age < 25:
            return 'Mature Adult\n(16-25 wks)'
        else:
            return 'Late Mature\n(>25 wks)'

    df['Age Cohort'] = df['Age at Dissection (wks)'].apply(bin_age)

   
    cohort_order = [
        'Young Adult\n(<16 wks)',
        'Mature Adult\n(16-25 wks)',
        'Late Mature\n(>25 wks)',
    ]
    cohort_summary = (
        df.groupby('Age Cohort')['Total Spaceflight (n)']
        .sum()
        .reindex(cohort_order)
        .fillna(0)
    )

    
    cohort_colors = [COLOR_GC, '#2563EB', COLOR_SF]

    bars = ax4.bar(
        cohort_summary.index,
        cohort_summary.values,
        color=cohort_colors,
        edgecolor='black',
        lw=0.7,
        width=0.45,
    )

 
    for bar in bars:
        height = bar.get_height()
        ax4.text(
            bar.get_x() + bar.get_width() / 2.0,
            height + 0.4,
            f'n={int(height)}',
            ha='center',
            va='bottom',
            weight='bold',
            fontsize=9.5,
            color='#1F2937',
        )

    ax4.set_ylabel(
        'Total Flown Samples (n)', fontsize=10, weight='bold'
    )
    ax4.set_title(
        'Spaceflight Sample Allocation by Biological Age Cohort',
        fontsize=11,
        weight='bold',
        loc='left',
        pad=12,
    )


    ax4.set_ylim(0, cohort_summary.max() + 4)

    ax4.spines['top'].set_visible(False)
    ax4.spines['right'].set_visible(False)
    ax4.yaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax4.set_axisbelow(True)

    plt.tight_layout()
    plt.savefig(
        'notebooks/figure4_age_cohort_distribution.png', bbox_inches='tight'
    )
    plt.close()

    return df


age_df = generate_age_visuals()
print(
    age_df[[
        'Dataset ID',
        'Mission',
        'Age at Launch (wks)',
        'Mission Duration (days)',
        'Age at Dissection (wks)',
    ]].to_string(index=False)
)
```

## Grouped panels of mice analysis figures

```python
import os
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec
import pandas as pd

plt.style.use('seaborn-v0_8-paper')
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['font.sans-serif'] = ['Arial', 'Helvetica', 'DejaVu Sans']
plt.rcParams['axes.linewidth'] = 1.0

COLOR_SF = '#1E3A8A'  # Deep Space Blue (Spaceflight)
COLOR_GC = '#0D9488'  # Slate Teal (Ground Control)
COLOR_MID = '#2563EB'  # Royal Blue (Intermediate Cohort)
COLOR_LINE = '#94A3B8'  # Slate Gray (Connector lines)


def generate_multipanel_visuals(df_matrix, df_age):
    os.makedirs('notebooks', exist_ok=True)

    
    # MULTI-PANEL FIGURE 1: Meta-Analysis Sample & Volume Matrix (1A & 1B)
    
    fig1 = plt.figure(figsize=(13, 5.2), dpi=300)
    gs1 = gridspec.GridSpec(1, 2, width_ratios=[1.2, 1.0], wspace=0.28)

    ax1a = fig1.add_subplot(gs1[0])
    ax1b = fig1.add_subplot(gs1[1])

    
    # Panel 1A: Stacked Sample Distribution by Dataset
    
    df = df_matrix.copy()
    df['Total Mice (n)'] = df['Spaceflight (SF)'] + df['Ground Control (GC)']
    df['Label'] = df['Dataset ID'] + '\n(' + df['Mission'] + ')'

    ax1a.bar(
        df['Label'],
        df['Spaceflight (SF)'],
        label='Spaceflight (SF)',
        color=COLOR_SF,
        edgecolor='black',
        lw=0.7,
        width=0.55,
    )
    ax1a.bar(
        df['Label'],
        df['Ground Control (GC)'],
        bottom=df['Spaceflight (SF)'],
        label='Ground Control (GC)',
        color=COLOR_GC,
        edgecolor='black',
        lw=0.7,
        width=0.55,
    )

    for i in range(len(df)):
        sf, gc, total = (
            df['Spaceflight (SF)'][i],
            df['Ground Control (GC)'][i],
            df['Total Mice (n)'][i],
        )
        ax1a.text(
            i,
            sf / 2,
            str(sf),
            ha='center',
            va='center',
            color='white',
            weight='bold',
            fontsize=8,
        )
        ax1a.text(
            i,
            sf + gc / 2,
            str(gc),
            ha='center',
            va='center',
            color='white',
            weight='bold',
            fontsize=8,
        )
        ax1a.text(
            i,
            total + 0.8,
            f'n={total}',
            ha='center',
            va='bottom',
            color='#1F2937',
            weight='bold',
            fontsize=8,
        )

    ax1a.set_ylabel(
        'Biological Replicates Count (n)', fontsize=9.5, weight='bold'
    )
    ax1a.set_title(
        'A', fontsize=14, weight='bold', loc='left', pad=10
    )  
    ax1a.set_title(
        'Sample Distribution Matrix (N=79)',
        fontsize=10.5,
        weight='bold',
        loc='center',
        pad=10,
    )
    ax1a.set_ylim(0, max(df['Total Mice (n)']) + 5)
    ax1a.spines['top'].set_visible(False)
    ax1a.spines['right'].set_visible(False)
    ax1a.yaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax1a.set_axisbelow(True)
    ax1a.legend(
        frameon=True,
        facecolor='white',
        framealpha=0.9,
        edgecolor='#E5E7EB',
        loc='upper right',
        fontsize=8,
    )

    
    # Panel 1B: Grouped Volume by Omics Layer
    
    omics_summary = df.groupby('Omics Layer')[
        ['Spaceflight (SF)', 'Ground Control (GC)']
    ].sum()
    x = range(len(omics_summary))
    width = 0.32

    b1 = ax1b.bar(
        [p - width / 2 for p in x],
        omics_summary['Spaceflight (SF)'],
        width,
        label='Spaceflight (SF)',
        color=COLOR_SF,
        edgecolor='black',
        lw=0.7,
    )
    b2 = ax1b.bar(
        [p + width / 2 for p in x],
        omics_summary['Ground Control (GC)'],
        width,
        label='Ground Control (GC)',
        color=COLOR_GC,
        edgecolor='black',
        lw=0.7,
    )

    ax1b.bar_label(
        b1, fmt='%d', padding=3, fontsize=8.5, weight='bold', color='#1F2937'
    )
    ax1b.bar_label(
        b2, fmt='%d', padding=3, fontsize=8.5, weight='bold', color='#1F2937'
    )

    ax1b.set_xticks(x)
    ax1b.set_xticklabels(omics_summary.index, fontsize=9, weight='bold')
    ax1b.set_ylabel(
        'Total Pooled Replicates (n)', fontsize=9.5, weight='bold'
    )
    ax1b.set_title(
        'B', fontsize=14, weight='bold', loc='left', pad=10
    )  # Subpanel Label
    ax1b.set_title(
        'Sample Volume by Omics Layer',
        fontsize=10.5,
        weight='bold',
        loc='center',
        pad=10,
    )
    ax1b.set_ylim(0, omics_summary.values.max() + 6)
    ax1b.spines['top'].set_visible(False)
    ax1b.spines['right'].set_visible(False)
    ax1b.yaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax1b.set_axisbelow(True)
    ax1b.legend(
        frameon=True,
        facecolor='white',
        framealpha=0.9,
        edgecolor='#E5E7EB',
        loc='lower right',
        fontsize=8,
    )

    plt.tight_layout()
    plt.savefig('notebooks/figure1_multipanel_matrix.png', bbox_inches='tight')
    plt.close()

    
    # MULTI-PANEL FIGURE 2: Rodent Lifespan & Cohort Profiling (2A & 2B)
    
    fig2 = plt.figure(figsize=(13, 5.2), dpi=300)
    gs2 = gridspec.GridSpec(1, 2, width_ratios=[1.2, 0.8], wspace=0.28)

    ax2a = fig2.add_subplot(gs2[0])
    ax2b = fig2.add_subplot(gs2[1])

    
    # Panel 2A: Lifespan Timeline
    
    dfa = df_age.copy()
    dfa['Label'] = dfa['Dataset ID'] + '\n(' + dfa['Mission'] + ')'

    for i in range(len(dfa)):
        launch_age = dfa['Age at Launch (wks)'][i]
        dissect_age = dfa['Age at Dissection (wks)'][i]

        ax2a.plot(
            [launch_age, dissect_age],
            [i, i],
            color=COLOR_LINE,
            lw=2.0,
            zorder=1,
        )
        ax2a.scatter(
            launch_age,
            i,
            color=COLOR_GC,
            s=45,
            label='Launch Baseline' if i == 0 else '',
            zorder=3,
        )
        ax2a.scatter(
            dissect_age,
            i,
            color=COLOR_SF,
            s=55,
            marker='X',
            label='Dissection Terminal' if i == 0 else '',
            zorder=3,
        )

        ax2a.text(
            launch_age - 0.6,
            i,
            f'{launch_age:.1f}',
            ha='right',
            va='center',
            fontsize=7.5,
            weight='bold',
            color='#374151',
        )
        ax2a.text(
            dissect_age + 0.6,
            i,
            f'{dissect_age:.1f}',
            ha='left',
            va='center',
            fontsize=7.5,
            weight='bold',
            color=COLOR_SF,
        )

    ax2a.set_yticks(range(len(dfa)))
    ax2a.set_yticklabels(dfa['Label'], fontsize=8)
    ax2a.set_xlabel(
        'Chronological Animal Age (Weeks)', fontsize=9.5, weight='bold'
    )
    ax2a.set_title(
        'A', fontsize=14, weight='bold', loc='left', pad=10
    )  # Subpanel Label
    ax2a.set_title(
        'Payload Lifespan Development Timeline',
        fontsize=10.5,
        weight='bold',
        loc='center',
        pad=10,
    )
    ax2a.set_xlim(
        dfa['Age at Launch (wks)'].min() - 3.0,
        dfa['Age at Dissection (wks)'].max() + 3.0,
    )
    ax2a.spines['top'].set_visible(False)
    ax2a.spines['right'].set_visible(False)
    ax2a.xaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax2a.set_axisbelow(True)
    ax2a.legend(
        frameon=True,
        facecolor='white',
        framealpha=0.9,
        edgecolor='#E5E7EB',
        loc='lower right',
        fontsize=8,
    )

    
    # Panel 2B: Corrected Biological Age Cohort Binning
    
    def bin_age_corrected(age):
        if age < 16:
            return 'Young Adult\n(<16 wks)'
        elif age < 25:
            return 'Mature Adult\n(16-25 wks)'
        else:
            return 'Late Mature\n(>25 wks)'  # Scientifically precise label

    dfa['Age Cohort'] = dfa['Age at Dissection (wks)'].apply(bin_age_corrected)
    cohort_order = [
        'Young Adult\n(<16 wks)',
        'Mature Adult\n(16-25 wks)',
        'Late Mature\n(>25 wks)',
    ]
    cohort_summary = (
        dfa.groupby('Age Cohort')['Total Spaceflight (n)']
        .sum()
        .reindex(cohort_order)
        .fillna(0)
    )

    bars = ax2b.bar(
        cohort_summary.index,
        cohort_summary.values,
        color=[COLOR_GC, COLOR_MID, COLOR_SF],
        edgecolor='black',
        lw=0.7,
        width=0.45,
    )

    for bar in bars:
        h = bar.get_height()
        ax2b.text(
            bar.get_x() + bar.get_width() / 2.0,
            h + 0.4,
            f'n={int(h)}',
            ha='center',
            va='bottom',
            weight='bold',
            fontsize=8.5,
            color='#1F2937',
        )

    ax2b.set_ylabel(
        'Total Flown Samples (n)', fontsize=9.5, weight='bold'
    )
    ax2b.set_title(
        'B', fontsize=14, weight='bold', loc='left', pad=10
    )  
    ax2b.set_title(
        'Sample Allocation by Age Cohort',
        fontsize=10.5,
        weight='bold',
        loc='center',
        pad=10,
    )
    ax2b.set_ylim(0, cohort_summary.max() + 4)
    ax2b.spines['top'].set_visible(False)
    ax2b.spines['right'].set_visible(False)
    ax2b.yaxis.grid(True, linestyle='--', alpha=0.4, zorder=0)
    ax2b.set_axisbelow(True)

    plt.tight_layout()
    plt.savefig('notebooks/figure2_multipanel_age.png', bbox_inches='tight')
    plt.close()



generate_multipanel_visuals(df, df_age)
```
